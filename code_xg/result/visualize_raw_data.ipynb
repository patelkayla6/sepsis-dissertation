## visualize the feature across time

```python
import pandas as pd

subj_id = '1017'
df_ecg_0 = pd.read_csv(f'../../data/ECG/003-{subj_id}_ecg_lead0_20s.csv')
df_ecg_1 = pd.read_csv(f'../../data/ECG/003-{subj_id}_ecg_lead1_20s.csv')
df_ppg = pd.read_csv(f'../../data/PPG/01NVa-003-{subj_id}_20s.csv')
```

```python
df_ecg_0
```

```python
df_ppg
```

```python
import matplotlib.pyplot as plt

# Plot ECG_a_R over time
plt.figure(figsize=(12, 6))
plt.plot(df_ecg_0['Time_Elapsed_Seconds']/60/60, df_ecg_0['HRV_Prc20NN'], label='ECG Lead 0', linewidth=2)
plt.plot(df_ecg_1['Time_Elapsed_Seconds']/60/60, df_ecg_1['HRV_Prc20NN'], label='ECG Lead 1', linewidth=2)
plt.plot(df_ppg['Time_Elapsed_Seconds']/60/60, df_ppg['HRV_Prc20NN'], label='PPG', linewidth=2)

plt.xlabel('Time (h)')
plt.ylabel('HRV_SD2d')
plt.title('HRV Over Time')
plt.legend()
plt.grid(True)
plt.show()
```
