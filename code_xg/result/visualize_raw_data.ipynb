```python
import pandas as pd
import os

# Subject IDs to exclude
excluded_ids = {'003-1011', '003-1016', '003-1017', '003-1214', '003-1219', '003-1239'}

# Directory path to PPG files
ppg_dir = '../../data/PPG/'

# Generate subject IDs to include
included_ids = [f'003-{i:04d}' for i in range(1001, 1260) if f'003-{i:04d}' not in excluded_ids]

# Prepare a list to store results
results = []

# Loop through all included subject IDs
for subj_id in included_ids:
    filename = f'01NVa-{subj_id}_20s.csv'
    filepath = os.path.join(ppg_dir, filename)

    if os.path.exists(filepath):
        try:
            df = pd.read_csv(filepath)
            if 'Time_Elapsed_Seconds' in df.columns:
                signal_length = df['Time_Elapsed_Seconds'].iloc[-1]
                results.append({'subject_id': subj_id, 'signal_length': signal_length})
            else:
                results.append({'subject_id': subj_id, 'signal_length': None, 'error': 'Column missing'})
        except Exception as e:
            results.append({'subject_id': subj_id, 'signal_length': None, 'error': str(e)})
    else:
        results.append({'subject_id': subj_id, 'signal_length': None, 'error': 'File not found'})

# Convert to DataFrame
df_results = pd.DataFrame(results)

# Save or display as needed
print(df_results)


# Drop None values
valid_lengths = df_results['signal_length'].dropna()

# Plot histogram
plt.figure(figsize=(10, 6))
plt.hist(valid_lengths/3600, bins=50, edgecolor='black')
plt.title('Histogram of Signal Lengths (Time_Elapsed_Seconds)')
plt.xlabel('Signal Length (hrs)')
plt.ylabel('Number of Recordings')
plt.grid(True)
plt.tight_layout()
plt.show()
```
