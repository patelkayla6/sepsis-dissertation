```python
import pandas as pd
import matplotlib.pyplot as plt
import math


# 3) List all feature columns
features = [c for c in merged.columns if c not in ['icustay_id', 'mortality']]

# 4) Set up grid
n_feats = len(features)
ncols = 4
nrows = math.ceil(n_feats / ncols)
fig, axes = plt.subplots(nrows, ncols, figsize=(ncols * 4, nrows * 3))
axes = axes.flatten()

# 5) Plot normalized histograms
for idx, feat in enumerate(features):
    ax = axes[idx]
    surv = merged[merged['mortality'] == 0][feat].dropna()
    died = merged[merged['mortality'] == 1][feat].dropna()

    ax.hist(surv, bins=15, density=True, alpha=0.6, label='Survived')
    ax.hist(died, bins=15, density=True, alpha=0.6, label='Died')

    ax.set_title(feat, fontsize=8)
    ax.set_ylabel('Density')
    ax.tick_params(axis='both', labelsize=6, direction='out')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    if idx == 0:
        ax.legend(fontsize=6, frameon=False)

# 6) Remove empty axes and show
for j in range(idx + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.metrics import roc_auc_score, classification_report

# 1) Prepare data
X = merged.drop(columns=['icustay_id', 'mortality'])
y = merged['mortality']

# 2) Fill NaNs with zero
X = X.fillna(0)

# 3) Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    stratify=y,
    random_state=42
)

# 4) Fit a baseline XGBClassifier
xgb_clf = xgb.XGBClassifier(
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)
xgb_clf.fit(X_train, y_train)

# 5) Evaluate baseline
probs = xgb_clf.predict_proba(X_test)[:, 1]
preds = xgb_clf.predict(X_test)
print("Baseline XGB ROC AUC:", roc_auc_score(y_test, probs))
print("\nBaseline classification report:\n",
      classification_report(y_test, preds, digits=3))

# 6) Quick hyperparameter search
param_dist = {
    'n_estimators':    [100, 200, 500],
    'max_depth':       [3, 6, 10, 20],
    'learning_rate':   [0.01, 0.1, 0.2, 0.3],
    'subsample':       [0.6, 0.8, 1.0],
    'colsample_bytree':[0.6, 0.8, 1.0],
    'gamma':           [0, 1, 5]
}

rs = RandomizedSearchCV(
    xgb.XGBClassifier(
        use_label_encoder=False,
        eval_metric='logloss',
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=param_dist,
    n_iter=20,
    scoring='roc_auc',
    cv=4,
    random_state=42,
    n_jobs=-1
)

rs.fit(X_train, y_train)

print("Best XGB params:", rs.best_params_)

# 7) Evaluate tuned model
best_xgb = rs.best_estimator_
best_probs = best_xgb.predict_proba(X_test)[:, 1]
best_preds = (best_probs > 0.5).astype(int)

print("Tuned XGB ROC AUC:", roc_auc_score(y_test, best_probs))
print("\nTuned classification report:\n",
      classification_report(y_test, best_preds, digits=3))
```

```python
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.metrics import roc_auc_score

# 1) Prepare X, y and fill NaNs with zero
X = merged.drop(columns=['icustay_id', 'mortality']).fillna(0)
y = merged['mortality']

# 2) Set up 10-fold stratified CV
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

# 3) Initialize XGBClassifier
clf = xgb.XGBClassifier(
    use_label_encoder=False,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

# 4) Compute AUC per fold
fold_aucs = cross_val_score(
    clf,
    X, y,
    cv=skf,
    scoring='roc_auc',
    n_jobs=-1
)
print("AUC for each fold:", fold_aucs)
print(f"Mean AUC: {fold_aucs.mean():.3f}  Std: {fold_aucs.std():.3f}")

# 5) Aggregate predictions over all test folds and compute overall AUC
probs = cross_val_predict(
    clf,
    X, y,
    cv=skf,
    method='predict_proba',
    n_jobs=-1
)[:, 1]
aggregated_auc = roc_auc_score(y, probs)
print("Aggregated 10-fold CV AUC:", aggregated_auc)
```

```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    recall_score,
    f1_score
)

# 1) Prepare X, y (keep NaNs)
X = merged.drop(columns=['icustay_id', 'mortality'])
y = merged['mortality']

# 2) Identify numeric vs categorical cols
numeric_cols     = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()

# 3) Build preprocessing pipeline
numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler',  StandardScaler()),
])
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot',  OneHotEncoder(drop='first', sparse=False, handle_unknown='ignore')),
])
preprocessor = ColumnTransformer([
    ('num', numeric_pipeline,   numeric_cols),
    ('cat', categorical_pipeline, categorical_cols),
])

# 4) Combine with RandomForest
base_clf = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier',    RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )),
])

# 5) Run 20 repeats of 10-fold CV, aggregating per-repeat metrics
results = []
for rep in range(20):
    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42+rep)

    # get out-of-fold probabilities and predictions
    probs = cross_val_predict(
        base_clf, X, y,
        cv=skf,
        method='predict_proba',
        n_jobs=-1
    )[:, 1]
    preds = (probs >= 0.5).astype(int)

    # compute metrics for this repeat
    results.append({
        'AUROC':       roc_auc_score(y, probs),
        'AUPRC':       average_precision_score(y, probs),
        'Sensitivity': recall_score(y, preds, pos_label=1),
        'Specificity': recall_score(y, preds, pos_label=0),
        'F1-score':    f1_score(y, preds),
    })

# 6) Aggregate over repeats
df = pd.DataFrame(results)
mean = df.mean()
std  = df.std()

# 7) Format as mean±std percentages
formatted = (
    (mean * 100).round(1).astype(str) + '%±' +
    (std  * 100).round(1).astype(str) + '%'
)
formatted.name = 'Average Metrics over 20×10-fold CV'

# 8) Display
print("Classifier: Random_Forest")
print(formatted)
```
